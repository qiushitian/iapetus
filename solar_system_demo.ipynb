## Initialize Notebook

```python
import amuse
from amuse.ext.solarsystem import new_solar_system
```

## Creating Solar System

```python
our_solar_system = new_solar_system()
```

But this solar system has all planets (even including Pluto). We can remove everything else so that there are only the Sun, Jupiter, and Saturn left.

```python
our_solar_system.remove_particles(our_solar_system[[1, 2, 3, 4, 7, 8, 9]])
```

```python
print(our_solar_system)
```

This `our_solar_system` is an instance of the `Particles` object.

```python
type(our_solar_system)
```

And each celestial body is apparently an instance of the `Particle` object.

```python
type(our_solar_system[1])
```
